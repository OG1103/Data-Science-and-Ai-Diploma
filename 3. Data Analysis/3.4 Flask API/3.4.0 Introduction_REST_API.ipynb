# 1. What is a REST API?

**REST** (Representational State Transfer) is an architectural style for designing networked applications. A **REST API** allows different applications to communicate over HTTP.

## Key Concepts

| Concept | Description |
|---------|-------------|
| **Client** | The application making requests (browser, mobile app, etc.) |
| **Server** | The application responding to requests (your API) |
| **Resource** | Data entities (users, products, orders) |
| **Endpoint** | URL path to access a resource (`/api/users`) |
| **HTTP Methods** | Actions to perform (GET, POST, PUT, DELETE) |

# 2. HTTP Methods (CRUD Operations)

| Method | CRUD | Description | Example |
|--------|------|-------------|---------|
| `GET` | Read | Retrieve resource(s) | Get all users, Get user by ID |
| `POST` | Create | Create new resource | Create new user |
| `PUT` | Update | Replace entire resource | Update user (all fields) |
| `PATCH` | Update | Update part of resource | Update user (some fields) |
| `DELETE` | Delete | Remove resource | Delete user |

## Example Endpoints

```
GET    /api/users          - Get all users
GET    /api/users/1        - Get user with ID 1
POST   /api/users          - Create new user
PUT    /api/users/1        - Update user with ID 1
DELETE /api/users/1        - Delete user with ID 1
```

# 3. HTTP Status Codes

Status codes tell the client what happened with their request.

## Success Codes (2xx)

| Code | Meaning | When to Use |
|------|---------|-------------|
| `200 OK` | Success | GET request successful |
| `201 Created` | Resource created | POST created new resource |
| `204 No Content` | Success, no body | DELETE successful |

## Client Error Codes (4xx)

| Code | Meaning | When to Use |
|------|---------|-------------|
| `400 Bad Request` | Invalid request | Malformed JSON, validation error |
| `401 Unauthorized` | Not authenticated | Missing or invalid credentials |
| `403 Forbidden` | Not authorized | Valid credentials, but no permission |
| `404 Not Found` | Resource not found | User with ID doesn't exist |
| `405 Method Not Allowed` | Wrong HTTP method | POST to read-only endpoint |
| `422 Unprocessable Entity` | Validation failed | Data didn't pass validation |

## Server Error Codes (5xx)

| Code | Meaning | When to Use |
|------|---------|-------------|
| `500 Internal Server Error` | Server error | Unhandled exception |
| `503 Service Unavailable` | Server overloaded | Maintenance, overload |

# 4. Request and Response Structure

## HTTP Request Components

```
POST /api/users HTTP/1.1              <- Method, URL, Protocol
Host: api.example.com                  <- Headers
Content-Type: application/json
Authorization: Bearer token123

{                                      <- Body (JSON)
    "name": "John Doe",
    "email": "john@example.com"
}
```

## HTTP Response Components

```
HTTP/1.1 201 Created                   <- Status line
Content-Type: application/json         <- Headers

{                                      <- Body (JSON)
    "id": 1,
    "name": "John Doe",
    "email": "john@example.com",
    "created_at": "2024-01-15T10:30:00Z"
}
```

# 5. JSON (JavaScript Object Notation)

**JSON** is the standard format for REST API data exchange.

## JSON Syntax

```json
{
    "string": "Hello World",
    "number": 42,
    "float": 3.14,
    "boolean": true,
    "null": null,
    "array": [1, 2, 3],
    "object": {
        "nested": "value"
    }
}
```

## Python JSON

````python
import json

# Python dict to JSON string
data = {
    "name": "John",
    "age": 30,
    "active": True
}

json_string = json.dumps(data, indent=2)
print("Python dict -> JSON string:")
print(json_string)

# JSON string to Python dict
json_text = '{"name": "Jane", "age": 25}'
parsed_data = json.loads(json_text)
print(f"\nJSON string -> Python dict: {parsed_data}")
print(f"Name: {parsed_data['name']}")
````

# 6. REST API Design Principles

## 6.1 Use Nouns, Not Verbs

```
GOOD:  GET  /api/users          (get users)
BAD:   GET  /api/getUsers       (verb in URL)

GOOD:  POST /api/users          (create user)
BAD:   POST /api/createUser     (verb in URL)
```

## 6.2 Use Plural Nouns

```
GOOD:  /api/users
BAD:   /api/user

GOOD:  /api/products
BAD:   /api/product
```

## 6.3 Use Hierarchical URLs for Relations

```
GET /api/users/1/orders         - Get orders for user 1
GET /api/users/1/orders/5       - Get order 5 for user 1
POST /api/users/1/orders        - Create order for user 1
```

## 6.4 Use Query Parameters for Filtering/Sorting

```
GET /api/users?status=active           - Filter by status
GET /api/users?sort=name&order=asc     - Sort by name
GET /api/users?page=2&limit=10         - Pagination
GET /api/products?category=electronics&min_price=100
```

# 7. What is Flask?

| Flask | Django |
|-------|--------|

# 9. Your First Flask App

Create a file called `app.py`:

````python
# app.py - Basic Flask application

code = '''
from flask import Flask

# Create Flask application
app = Flask(__name__)

# Define a route
@app.route('/')
def home():
    return 'Hello, World!'

@app.route('/api/hello')
def hello_api():
    return {'message': 'Hello from API!'}

# Run the app
if __name__ == '__main__':
    app.run(debug=True)
'''

print("Basic Flask App (save as app.py):")
print("="*50)
print(code)
````

## curl Examples

```bash
# GET request
curl http://localhost:5000/api/users

# GET with headers
curl -H "Authorization: Bearer token123" http://localhost:5000/api/users

# POST with JSON data
curl -X POST http://localhost:5000/api/users \
     -H "Content-Type: application/json" \
     -d '{"name": "John", "email": "john@email.com"}'

# PUT request
curl -X PUT http://localhost:5000/api/users/1 \
     -H "Content-Type: application/json" \
     -d '{"name": "John Updated"}'

# DELETE request
curl -X DELETE http://localhost:5000/api/users/1
```

````python
# Python requests library example
# First: pip install requests

import requests

# Note: These examples assume a Flask server is running

example_code = '''
import requests

BASE_URL = "http://localhost:5000/api"

# GET request
response = requests.get(f"{BASE_URL}/users")
print(response.status_code)  # 200
print(response.json())       # JSON data

# POST request
new_user = {"name": "John", "email": "john@email.com"}
response = requests.post(f"{BASE_URL}/users", json=new_user)
print(response.status_code)  # 201
print(response.json())

# PUT request
updated_user = {"name": "John Updated"}
response = requests.put(f"{BASE_URL}/users/1", json=updated_user)
print(response.json())

# DELETE request
response = requests.delete(f"{BASE_URL}/users/1")
print(response.status_code)  # 204

# With headers
headers = {"Authorization": "Bearer token123"}
response = requests.get(f"{BASE_URL}/users", headers=headers)
'''

print("Python requests examples:")
print(example_code)
````
