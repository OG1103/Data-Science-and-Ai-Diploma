# Matplotlib - The Foundation of Python Visualization

## 1. What is Matplotlib

Matplotlib is the most widely used Python library for creating static, animated, and interactive visualizations. It provides a MATLAB-like interface and gives you complete control over every aspect of your plots.

**Key Features:**
- Foundation for most Python visualization libraries
- Highly customizable (control every element of a plot)
- Works well with NumPy arrays and Pandas DataFrames
- Can create publication-quality figures
- Supports multiple output formats (PNG, PDF, SVG, etc.)

**Common Use Cases:**
- Line plots for trends over time
- Bar charts for categorical comparisons
- Scatter plots for relationships between variables
- Histograms for distribution analysis
- Subplots for multiple visualizations in one figure

## 2. Importing Matplotlib

- `import matplotlib.pyplot as plt` - Import Matplotlib's plotting interface

**Note:** Matplotlib is typically imported as `plt`

````python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# For displaying plots in Jupyter notebooks
%matplotlib inline
````

## 3. Basic Plot Types in Matplotlib

### 3.1 Line Plot

A line plot displays data points connected by lines, ideal for showing trends over time or continuous data.

**Syntax:**
- `plt.plot(x, y)` - Create a line plot
- `plt.xlabel('label')` - Set x-axis label
- `plt.ylabel('label')` - Set y-axis label
- `plt.title('title')` - Set plot title
- `plt.legend()` - Show legend (Box that identifies lines using set label in the plot paramter), Without plt.legend(), your labels won't be displayed even if you set them in the plot commands!
- `plt.show()` - Display the plot

**Common Parameters:**
- `color='red'` - Line color
- `linewidth=2` - Line thickness
- `linestyle='--'` - Line style (solid (syntax: "-"), dashed (syntax: "--"), dotted (syntax ":"))
- `marker='o'` - Marker style for data points
- `label='Name'` - Label for legend

````python
# Basic Line Plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Line Plot')
plt.show()

# Multiple Lines with Customization
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 6))  # Set figure size
plt.plot(x, y1, color='blue', linewidth=2, linestyle='-', marker='o', markersize=3, label='sin(x)') # First Line in plot with label sin(x) and color blue
plt.plot(x, y2, color='red', linewidth=2, linestyle='--', marker='s', markersize=3, label='cos(x)') # Second Line in plot with label cos(x) and color red
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Sine and Cosine Waves')
plt.legend()
plt.grid(True)  # Add grid
plt.show()
````

### 3.2 Bar Chart

Bar charts display categorical data with rectangular bars. Heights represent values, making comparisons easy.

**Syntax:**
- `plt.bar(x, height)` - Vertical bar chart
- `plt.barh(y, width)` - Horizontal bar chart

**Common Parameters:**
- `color='blue'` - Bar color
- `width=0.5` - Bar width
- `edgecolor='black'` - Border color
- `alpha=0.7` - Transparency (0-1)

````python
# Simple Bar Chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [25, 40, 30, 55, 45]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color='skyblue', edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart Example')
plt.show()

# Grouped Bar Chart
categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [20, 35, 30, 35]
product_b = [25, 32, 34, 20]

x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, product_a, width, label='Product A', color='#1f77b4')
plt.bar(x + width/2, product_b, width, label='Product B', color='#ff7f0e')
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.title('Quarterly Sales Comparison')
plt.xticks(x, categories)
plt.legend()
plt.show()

categories = ['A', 'B', 'C', 'D', 'E']
values = [25, 40, 30, 55, 45]

# Horizontal Bar Chart
plt.figure(figsize=(8, 5))
plt.barh(categories, values, color='lightcoral')
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title('Horizontal Bar Chart')
plt.show()
````

### 3.3 Scatter Plot

Scatter plots show the relationship between two numerical variables using dots. Each dot represents one observation. An observation is one data point - an x value paired with its corresponding y value.

**Syntax:**
- `plt.scatter(x, y)` - Create scatter plot

**Common Parameters:**
- `s=50` - Size of markers (can be a single value or array)
- `c='red'` - Color of markers (can be a single color, array of colors, or array of values for gradient)
- `marker='o'` - Marker shape ('o', 's', '^', 'D', etc.)
- `alpha=0.5` - Transparency (0=fully transparent, 1=fully opaque)
- `cmap='viridis'` - Color map for gradient coloring ('viridis', 'coolwarm', 'plasma', etc.)
- `edgecolors='black'` - Border color around markers
- `linewidth=1` - Width of marker borders

**Advanced Usage - Multiple Variables:**

Scatter plots can represent up to 4 variables simultaneously:
- **x-position**: First variable
- **y-position**: Second variable  
- **size (s parameter)**: Third variable (use array of values)
- **color (c parameter)**: Fourth variable (use array of values with cmap)

**Example with Multiple Variables:**
```python
# All arrays must have the same length
x = [7, 14, 9, 18, 12]           # x-position
y = [8, 5, 9, 6, 7]              # y-position
sizes = [500, 200, 800, 160, 300] # dot sizes (variable 3)
colors = [170, 40, 165, 38, 160]  # color values (variable 4)

plt.scatter(x, y, s=sizes, c=colors, cmap='coolwarm')
plt.colorbar(label='Color Scale')  # Show color gradient legend
```

**How Arrays Map to Points:**
- Each index position creates one complete dot
- Point at index 0: x[0], y[0], size=sizes[0], color=colors[0]
- Point at index 1: x[1], y[1], size=sizes[1], color=colors[1]
- And so on...

````python
# Basic Scatter Plot
x = np.random.rand(50)
y = np.random.rand(50)

plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=100, c='blue', alpha=0.5, edgecolors='black')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Basic Scatter Plot')
plt.show()

# Scatter Plot with Color Gradient
x = np.random.rand(100)
y = np.random.rand(100)
colors = np.random.rand(100)
sizes = 1000 * np.random.rand(100)

plt.figure(figsize=(10, 6))
scatter = plt.scatter(x, y, s=sizes, c=colors, cmap='viridis', alpha=0.6, edgecolors='black')
plt.colorbar(scatter, label='Color Scale')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter Plot with Size and Color Variations')
plt.show()
````

### 3.4 Histogram

Histograms show the distribution of numerical data by dividing it into bins (ranges) and counting observations in each bin.

**Syntax:**
- `plt.hist(data, bins=10)` - Create histogram

**Common Parameters:**
- `bins=20` - Number of bins/bars
- `color='green'` - Bar color
- `edgecolor='black'` - Border color
- `alpha=0.7` - Transparency
- `density=True` - Normalize to show probability density

````python
# Basic Histogram
data = np.random.randn(1000)  # 1000 random values from normal distribution

plt.figure(figsize=(10, 6))
plt.hist(data, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Random Data')
plt.grid(axis='y', alpha=0.5)
plt.show()

# Multiple Histograms
data1 = np.random.normal(100, 10, 1000)
data2 = np.random.normal(90, 20, 1000)

plt.figure(figsize=(10, 6))
plt.hist(data1, bins=30, alpha=0.5, label='Group A', color='blue')
plt.hist(data2, bins=30, alpha=0.5, label='Group B', color='red')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Overlapping Histograms')
plt.legend()
plt.show()
````

### 3.5 Pie Chart

Pie charts show proportions of a whole, with each slice representing a category's percentage.

**Syntax:**
- `plt.pie(sizes, labels=labels)` - Create pie chart

**Common Parameters:**
- `autopct='%1.1f%%'` - Show percentages
- `startangle=90` - Rotate chart
- `explode=[0.1, 0, 0]` - Separate slices
- `colors=['red', 'blue']` - Custom colors

````python
# Basic Pie Chart
sizes = [30, 25, 20, 15, 10]
labels = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc']

plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Market Share Distribution')
plt.axis('equal')  # Equal aspect ratio ensures pie is circular
plt.show()

# Pie Chart with Exploded Slice
explode = (0.1, 0, 0, 0, 0)  # Explode the 1st slice

plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, explode=explode, autopct='%1.1f%%', 
        startangle=90, shadow=True)
plt.title('Exploded Pie Chart')
plt.axis('equal')
plt.show()
````

### 3.6 Subplots - Multiple Plots

Subplots allow you to display multiple plots in a single figure, arranged in a grid layout.

**Syntax:**
- `plt.subplot(rows, cols, index)` - Create subplot at position
- `fig, axes = plt.subplots(rows, cols)` - Create figure with multiple subplots

**How it works:**
- `plt.subplot(2, 2, 1)` means: 2 rows, 2 columns, plot at position 1 (top-left)
- Positions are numbered left-to-right, top-to-bottom

````python
# Method 1: Using plt.subplot()
x = np.linspace(0, 10, 100)

plt.figure(figsize=(12, 8))

# First subplot (top-left)
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x), 'r')
plt.title('Sine Wave')

# Second subplot (top-right)
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'b')
plt.title('Cosine Wave')

# Third subplot (bottom-left)
plt.subplot(2, 2, 3)
plt.plot(x, np.tan(x), 'g')
plt.title('Tangent Wave')

# Fourth subplot (bottom-right)
plt.subplot(2, 2, 4)
plt.plot(x, x**2, 'purple')
plt.title('Quadratic')

plt.tight_layout()  # Adjust spacing between subplots
plt.show()

# Method 2: Using plt.subplots() - More flexible
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Access each subplot using axes[row, col]
axes[0, 0].plot(x, np.sin(x), 'r')
axes[0, 0].set_title('Sine Wave')

axes[0, 1].plot(x, np.cos(x), 'b')
axes[0, 1].set_title('Cosine Wave')

axes[1, 0].plot(x, np.tan(x), 'g')
axes[1, 0].set_title('Tangent Wave')

axes[1, 1].plot(x, x**2, 'purple')
axes[1, 1].set_title('Quadratic')

plt.tight_layout()
plt.show()
````
