# 1. What is MySQL?

MySQL is a **Relational Database Management System (RDBMS)** that uses **SQL (Structured Query Language)** to manage and manipulate data. It stores data in structured tables with rows and columns.

## Key Characteristics
- **Open-source** and widely used
- **ACID compliant** (Atomicity, Consistency, Isolation, Durability)
- **Relational** - data is organized in tables with relationships
- **Scalable** - handles small to large applications

## When to Use MySQL
- Structured data with clear relationships
- Applications requiring complex queries and joins
- Financial systems, e-commerce, user management
- When data integrity is critical

# 2. Database Concepts

## 2.1 Database Structure Hierarchy

```
Database Server (MySQL)
    └── Database (e.g., "ecommerce_db")
            └── Table (e.g., "users")
                    └── Columns (e.g., id, name, email)
                    └── Rows (actual data records)
```

## 2.2 Key Terms

| Term | Definition | Example |
|------|------------|----------|
| **Database** | Container for all tables | `ecommerce_db` |
| **Table** | Collection of related data | `users`, `products`, `orders` |
| **Column/Field** | Attribute of the data | `name`, `email`, `price` |
| **Row/Record** | Single entry in a table | One user's complete data |
| **Primary Key (PK)** | Unique identifier for each row | `user_id`, `product_id` |
| **Foreign Key (FK)** | Reference to another table's PK | `user_id` in orders table |
| **Schema** | Blueprint/structure of database | Table definitions, constraints |

# 4. Keys and Constraints

## 4.1 Primary Key (PK)

- **Uniquely identifies** each record in a table
- **Cannot be NULL**
- **Only one** primary key per table
- Usually auto-incremented integer

```sql
CREATE TABLE users (
    user_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100)
);
```

## 4.2 Foreign Key (FK)

- **Creates relationship** between two tables
- **References** a primary key in another table
- **Ensures referential integrity** (can't add invalid references)

```sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY AUTO_INCREMENT,
    user_id INT,
    FOREIGN KEY (user_id) REFERENCES users(user_id)
);
```

## 4.3 Other Constraints

| Constraint | Description | Example |
|------------|-------------|----------|
| `NOT NULL` | Value required | `email VARCHAR(100) NOT NULL` |
| `UNIQUE` | No duplicate values | `email VARCHAR(100) UNIQUE` |
| `DEFAULT` | Default value if not provided | `status VARCHAR(20) DEFAULT 'active'` |
| `CHECK` | Validates value | `CHECK (age >= 18)` |
| `INDEX` | Speeds up searches | `INDEX (email)` |

# 5. Entity Relationship Diagram (ERD)

## 5.1 What is an ERD?

An **Entity Relationship Diagram** is a visual representation of database tables and their relationships. It helps in:
- Planning database structure before implementation
- Communicating design with team members
- Identifying relationships and constraints

## 5.2 ERD Components

```
+-----------------+          +-----------------+
|     USERS       |          |     ORDERS      |
+-----------------+          +-----------------+
| PK: user_id     |---\      | PK: order_id    |
|     name        |    \---->| FK: user_id     |
|     email       |          |     total       |
|     created_at  |          |     created_at  |
+-----------------+          +-----------------+
```

### Symbols:
- **Rectangle**: Entity (table)
- **Line**: Relationship
- **PK**: Primary Key
- **FK**: Foreign Key

# 6. Relationships Types

## 6.1 One-to-One (1:1)

One record in Table A relates to exactly one record in Table B.

**Example**: User and User Profile

```
+----------+          +----------------+
|  USERS   |          |  USER_PROFILES |
+----------+          +----------------+
| user_id  |<-------->| profile_id     |
| name     |          | user_id (FK,UQ)|
+----------+          | bio            |
                      | avatar         |
                      +----------------+
```

```sql
CREATE TABLE users (
    user_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100)
);

CREATE TABLE user_profiles (
    profile_id INT PRIMARY KEY AUTO_INCREMENT,
    user_id INT UNIQUE,  -- UNIQUE ensures 1:1
    bio TEXT,
    avatar VARCHAR(255),
    FOREIGN KEY (user_id) REFERENCES users(user_id)
);
```

## 6.2 One-to-Many (1:N)

One record in Table A can relate to many records in Table B.

**Example**: User and Orders (one user can have many orders)

```
+----------+          +----------+
|  USERS   |          |  ORDERS  |
+----------+          +----------+
| user_id  |<---------| order_id |
| name     |     |----| user_id  | (FK)
+----------+     |    | total    |
                 |----| ...      |
                      +----------+
```

```sql
CREATE TABLE users (
    user_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100)
);

CREATE TABLE orders (
    order_id INT PRIMARY KEY AUTO_INCREMENT,
    user_id INT,  -- No UNIQUE, allows multiple orders per user
    total DECIMAL(10,2),
    FOREIGN KEY (user_id) REFERENCES users(user_id)
);
```

## 6.3 Many-to-Many (M:N)

Multiple records in Table A can relate to multiple records in Table B.

**Example**: Students and Courses (students can enroll in many courses, courses have many students)

**Requires a Junction/Bridge Table:**

```
+----------+     +----------------+     +----------+
| STUDENTS |     | ENROLLMENTS    |     | COURSES  |
+----------+     +----------------+     +----------+
| stud_id  |<----| enrollment_id  |---->| course_id|
| name     |     | student_id (FK)|     | title    |
+----------+     | course_id (FK) |     +----------+
                 | enrolled_date  |
                 +----------------+
```

```sql
CREATE TABLE students (
    student_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100)
);

CREATE TABLE courses (
    course_id INT PRIMARY KEY AUTO_INCREMENT,
    title VARCHAR(100)
);

-- Junction/Bridge Table
CREATE TABLE enrollments (
    enrollment_id INT PRIMARY KEY AUTO_INCREMENT,
    student_id INT,
    course_id INT,
    enrolled_date DATE,
    FOREIGN KEY (student_id) REFERENCES students(student_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id),
    UNIQUE (student_id, course_id)  -- Prevent duplicate enrollments
);
```

# 7. Normalization

**Normalization** is the process of organizing database tables to reduce redundancy and improve data integrity.

## 7.1 First Normal Form (1NF)

**Rules:**
- Each column contains atomic (indivisible) values
- No repeating groups or arrays

**Bad (Not 1NF):**
| user_id | name | phones |
|---------|------|--------|
| 1 | John | 123-456, 789-012 |

**Good (1NF):**
| user_id | name | phone |
|---------|------|-------|
| 1 | John | 123-456 |
| 1 | John | 789-012 |

## 7.2 Second Normal Form (2NF)

**Rules:**
- Must be in 1NF
- All non-key columns depend on the entire primary key

**Bad (Not 2NF):**
| order_id | product_id | product_name | quantity |
|----------|------------|--------------|----------|
| 1 | 101 | Laptop | 2 |

(product_name depends only on product_id, not full key)

**Good (2NF):** Split into two tables

## 7.3 Third Normal Form (3NF)

**Rules:**
- Must be in 2NF
- No transitive dependencies (non-key columns shouldn't depend on other non-key columns)

**Bad (Not 3NF):**
| emp_id | dept_id | dept_name |
|--------|---------|------------|
| 1 | 10 | Sales |

(dept_name depends on dept_id, not emp_id)

**Good (3NF):** Create separate departments table

# 8. Complete ERD Example: E-commerce Database

```
+------------------+       +------------------+       +------------------+
|      USERS       |       |      ORDERS      |       |   ORDER_ITEMS    |
+------------------+       +------------------+       +------------------+
| PK: user_id      |<------| PK: order_id     |<------| PK: item_id      |
|     username     |       | FK: user_id      |       | FK: order_id     |
|     email        |       |     total        |       | FK: product_id   |
|     password     |       |     status       |       |     quantity     |
|     created_at   |       |     created_at   |       |     price        |
+------------------+       +------------------+       +------------------+
                                                              |
                                                              |
+------------------+       +------------------+               |
|    CATEGORIES    |       |    PRODUCTS      |<--------------+
+------------------+       +------------------+
| PK: category_id  |<------| PK: product_id   |
|     name         |       | FK: category_id  |
|     description  |       |     name         |
+------------------+       |     price        |
                           |     stock        |
                           +------------------+
```

## Relationships Explained:
- **Users → Orders**: One-to-Many (1 user can have many orders)
- **Orders → Order_Items**: One-to-Many (1 order can have many items)
- **Products → Order_Items**: One-to-Many (1 product can be in many order items)
- **Categories → Products**: One-to-Many (1 category can have many products)

# 9. SQL to Create E-commerce Database

```sql
-- Create Database
CREATE DATABASE ecommerce_db;
USE ecommerce_db;

-- Users Table
CREATE TABLE users (
    user_id INT PRIMARY KEY AUTO_INCREMENT,
    username VARCHAR(50) NOT NULL UNIQUE,
    email VARCHAR(100) NOT NULL UNIQUE,
    password VARCHAR(255) NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Categories Table
CREATE TABLE categories (
    category_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50) NOT NULL,
    description TEXT
);

-- Products Table
CREATE TABLE products (
    product_id INT PRIMARY KEY AUTO_INCREMENT,
    category_id INT,
    name VARCHAR(100) NOT NULL,
    price DECIMAL(10, 2) NOT NULL,
    stock INT DEFAULT 0,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (category_id) REFERENCES categories(category_id)
);

-- Orders Table
CREATE TABLE orders (
    order_id INT PRIMARY KEY AUTO_INCREMENT,
    user_id INT NOT NULL,
    total DECIMAL(10, 2) NOT NULL,
    status ENUM('pending', 'processing', 'shipped', 'delivered') DEFAULT 'pending',
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (user_id) REFERENCES users(user_id)
);

-- Order Items Table (Junction for Orders-Products)
CREATE TABLE order_items (
    item_id INT PRIMARY KEY AUTO_INCREMENT,
    order_id INT NOT NULL,
    product_id INT NOT NULL,
    quantity INT NOT NULL,
    price DECIMAL(10, 2) NOT NULL,
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
```

# 10. Summary

## Key Takeaways

| Concept | Description |
|---------|-------------|
| **RDBMS** | Relational Database Management System |
| **Primary Key** | Unique identifier for each row |
| **Foreign Key** | Links tables together |
| **1:1 Relationship** | One record to one record |
| **1:N Relationship** | One record to many records |
| **M:N Relationship** | Many to many (needs junction table) |
| **Normalization** | Organizing data to reduce redundancy |
| **ERD** | Visual diagram of database structure |

## Best Practices

1. **Always use PRIMARY KEYs** - Every table should have one
2. **Use appropriate data types** - Don't use VARCHAR for numbers
3. **Name consistently** - Use snake_case, be descriptive
4. **Normalize to 3NF** - For most applications
5. **Plan with ERD first** - Before writing SQL
6. **Add indexes** - On frequently searched columns
7. **Use foreign keys** - Maintain data integrity
