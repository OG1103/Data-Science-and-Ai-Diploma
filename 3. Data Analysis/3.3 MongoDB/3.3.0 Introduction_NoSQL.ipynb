# 1. What is MongoDB?

**MongoDB** is a **NoSQL document database** that stores data in flexible, JSON-like documents called **BSON** (Binary JSON).

## Key Characteristics
- **Document-oriented**: Data stored as documents (not tables/rows)
- **Schema-flexible**: Documents can have different structures
- **Scalable**: Built for horizontal scaling (sharding)
- **High performance**: Optimized for read/write operations

## SQL vs NoSQL Comparison

| SQL (MySQL) | NoSQL (MongoDB) |
|-------------|----------------|
| Database | Database |
| Table | Collection |
| Row | Document |
| Column | Field |
| Primary Key | _id (auto-generated) |
| Foreign Key | Reference / Embedding |
| JOIN | $lookup / Embedding |

# 2. When to Use MongoDB vs MySQL

## Use MongoDB When:
- Data structure varies or evolves frequently
- Need to store hierarchical/nested data
- High write throughput required
- Horizontal scaling is needed
- Rapid prototyping/development
- Working with JSON/document data

**Examples**: Content management, user profiles, IoT data, logs, catalogs

## Use MySQL When:
- Data has clear, fixed structure
- Complex relationships between entities
- ACID transactions are critical
- Complex queries with multiple JOINs
- Data integrity is paramount

**Examples**: Financial systems, inventory, ERP, traditional business applications

# 3. MongoDB Data Structure

## 3.1 Document Structure

Documents are stored in **BSON** format (similar to JSON):

```json
{
    "_id": ObjectId("507f1f77bcf86cd799439011"),
    "name": "John Doe",
    "email": "john@email.com",
    "age": 30,
    "active": true,
    "tags": ["developer", "python"],
    "address": {
        "street": "123 Main St",
        "city": "New York",
        "zip": "10001"
    },
    "created_at": ISODate("2024-01-15T10:30:00Z")
}
```

## 3.2 Key Features

| Feature | Description |
|---------|-------------|
| **_id** | Unique identifier (auto-generated ObjectId) |
| **Nested Objects** | Documents within documents |
| **Arrays** | Lists of values or documents |
| **Flexible Schema** | Different documents can have different fields |

## 3.3 BSON Data Types

| Type | Description | Example |
|------|-------------|----------|
| `String` | UTF-8 text | `"Hello"` |
| `Integer` | 32-bit or 64-bit integer | `42` |
| `Double` | 64-bit floating point | `3.14` |
| `Boolean` | True or False | `true` |
| `Date` | Milliseconds since epoch | `ISODate()` |
| `ObjectId` | 12-byte unique identifier | `ObjectId("...")` |
| `Array` | Ordered list | `[1, 2, 3]` |
| `Object` | Embedded document | `{"key": "value"}` |
| `Null` | Null value | `null` |
| `Binary` | Binary data | Binary files |
| `Decimal128` | High-precision decimal | Financial data |

# 5. Data Modeling Patterns

## 5.1 Embedding (Denormalization)

Store related data **within the same document**.

```json
// User with embedded addresses
{
    "_id": ObjectId("..."),
    "name": "John Doe",
    "addresses": [
        {"type": "home", "street": "123 Main St", "city": "NYC"},
        {"type": "work", "street": "456 Office Rd", "city": "NYC"}
    ]
}
```

**Pros:**
- Single query retrieves all data
- No joins needed
- Better read performance

**Cons:**
- Data duplication
- Document size limit (16MB)
- Updates can be complex

**Use When:**
- Data is always accessed together
- One-to-few relationship
- Data doesn't change frequently

## 5.2 Referencing (Normalization)

Store related data in **separate documents** with references.

```json
// Users collection
{
    "_id": ObjectId("user123"),
    "name": "John Doe",
    "email": "john@email.com"
}

// Orders collection (references user)
{
    "_id": ObjectId("order456"),
    "user_id": ObjectId("user123"),  // Reference to user
    "total": 99.99,
    "items": [...]
}
```

**Pros:**
- No data duplication
- Smaller documents
- Easier updates

**Cons:**
- Multiple queries needed
- Requires $lookup for joins

**Use When:**
- One-to-many or many-to-many relationships
- Data accessed independently
- Data changes frequently

## 5.3 Schema Design Examples

### Blog Application

```json
// Posts collection - Embed comments (few comments per post)
{
    "_id": ObjectId("..."),
    "title": "My First Post",
    "content": "Hello World...",
    "author": {
        "_id": ObjectId("user123"),
        "name": "John Doe"
    },
    "comments": [
        {"user": "Jane", "text": "Great post!", "date": ISODate("...")},
        {"user": "Bob", "text": "Thanks!", "date": ISODate("...")}
    ],
    "tags": ["python", "mongodb"],
    "created_at": ISODate("...")
}
```

### E-commerce Application

```json
// Products collection
{
    "_id": ObjectId("..."),
    "name": "Laptop",
    "price": 999.99,
    "category_id": ObjectId("cat123"),  // Reference to category
    "specifications": {
        "cpu": "Intel i7",
        "ram": "16GB",
        "storage": "512GB SSD"
    },
    "reviews": []  // Could embed or reference based on volume
}

// Orders collection
{
    "_id": ObjectId("..."),
    "user_id": ObjectId("user123"),
    "items": [
        {"product_id": ObjectId("..."), "name": "Laptop", "price": 999.99, "qty": 1}
    ],
    "total": 999.99,
    "status": "shipped"
}
```

# 7. MongoDB Shell Commands

Basic commands to use in MongoDB Shell (`mongosh`):

```javascript
// Show all databases
show dbs

// Switch to a database (creates if doesn't exist)
use myapp

// Show collections in current database
show collections

// Insert a document
db.users.insertOne({name: "John", email: "john@email.com"})

// Find all documents
db.users.find()

// Find with condition
db.users.find({name: "John"})

// Pretty print
db.users.find().pretty()

// Count documents
db.users.countDocuments()

// Drop collection
db.users.drop()

// Drop database
db.dropDatabase()
```

# 8. Summary

## Key Concepts

| Concept | Description |
|---------|-------------|
| **Document** | JSON-like data structure (BSON) |
| **Collection** | Group of documents (like a table) |
| **Database** | Container for collections |
| **_id** | Unique identifier for each document |
| **Embedding** | Store related data in same document |
| **Referencing** | Store references to other documents |

## MongoDB vs MySQL Summary

| Feature | MongoDB | MySQL |
|---------|---------|-------|
| Schema | Flexible | Fixed |
| Data Format | Documents (JSON) | Tables (Rows) |
| Relationships | Embed or Reference | Foreign Keys + JOINs |
| Scaling | Horizontal (Sharding) | Vertical |
| Transactions | Supported (4.0+) | Strong ACID |
| Best For | Flexible data, rapid dev | Structured data, complex queries |
