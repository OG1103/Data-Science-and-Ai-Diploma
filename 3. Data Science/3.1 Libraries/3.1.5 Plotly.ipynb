# Plotly - Interactive Data Visualization

## 1. What is Plotly

Plotly is a Python library for creating interactive, publication-quality graphs. Unlike Matplotlib and Seaborn (which create static images), Plotly creates web-based interactive visualizations.

**Key Features:**
- **Interactive** - Hover, zoom, pan, click to explore data
- **Web-based** - Can be embedded in web applications
- **Publication quality** - Professional-looking plots
- **3D support** - Create 3D visualizations
- **Dashboards** - Build interactive dashboards with Dash

**Two Interfaces:**
1. **Plotly Express (px)** - High-level, simple, quick plots (recommended for most users)
2. **Graph Objects (go)** - Low-level, more control, complex customization

**When to use Plotly:**
- When you need interactive visualizations
- For web-based dashboards and applications
- When presenting data to non-technical audiences
- For exploratory data analysis with zoom/pan capabilities

## 2. Importing Plotly

- `import plotly.express as px` - Import Plotly Express (high-level interface)
- `import plotly.graph_objects as go` - Import Plotly Graph Objects (low-level interface)

**Note:** Plotly Express is typically imported as `px`

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
```

## 3. Plotly Express - Quick Interactive Plots

### 3.1 Interactive Scatter Plot

**Syntax:**
- `px.scatter(df, x='col1', y='col2')` - Create interactive scatter plot

**Common Parameters:**
- `color='category'` - Color points by category
- `size='value'` - Size points by value
- `hover_data=['col1', 'col2']` - Show extra info on hover
- `title='Title'` - Set plot title

```python
# Load sample data
iris = px.data.iris()

# Basic Interactive Scatter
fig = px.scatter(iris, x='sepal_width', y='sepal_length', 
                 color='species', 
                 size='petal_length',
                 hover_data=['petal_width'],
                 title='Iris Dataset - Interactive Scatter Plot')
fig.show()

# Scatter with custom styling
tips_df = px.data.tips()
fig = px.scatter(tips_df, x='total_bill', y='tip', 
                 color='time', size='size',
                 title='Tips Dataset',
                 labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'}, # Replace total_bill and tip with replaced string everywhere in the graph
                 template='plotly_dark')  # Dark theme
fig.show()
```

### 3.2 Interactive Line Plot

**Syntax:**
- `px.line(df, x='col1', y='col2')` - Create interactive line plot

```python
# Simple line plot
df = px.data.gapminder().query("country=='Canada'")
fig = px.line(df, x='year', y='lifeExp', 
              title='Life Expectancy in Canada Over Time')
fig.show()

# Multiple lines
df = px.data.gapminder().query("continent=='Europe'")
fig = px.line(df, x='year', y='lifeExp', color='country',
              title='Life Expectancy in European Countries')
fig.show()
```

### 3.3 Interactive Bar Chart

**Syntax:**
- `px.bar(df, x='category', y='value')` - Create interactive bar chart

```python
# Simple bar chart
data = px.data.tips()
fig = px.bar(data, x='day', y='total_bill', 
             color='sex',
             title='Total Bills by Day and Gender',
             barmode='group')  # 'group' for grouped, 'stack' for stacked
fig.show()

# Horizontal bar chart
fig = px.bar(data, y='day', x='total_bill', 
             orientation='h',
             color='time',
             title='Horizontal Bar Chart')
fig.show()
```

### 3.4 Interactive Histogram and Box Plot

**Syntax:**
- `px.histogram(df, x='column')` - Interactive histogram
- `px.box(df, x='category', y='value')` - Interactive box plot

```python
# Histogram
fig = px.histogram(tips_df, x='total_bill', 
                   nbins=20,
                   color='sex',
                   title='Distribution of Total Bill')
fig.show()

# Box Plot
fig = px.box(tips_df, x='day', y='total_bill', 
             color='time',
             title='Total Bill Distribution by Day')
fig.show()

# Violin Plot (like Seaborn)
fig = px.violin(tips_df, x='day', y='total_bill',
                color='sex',
                box=True,  # Show box plot inside
                title='Violin Plot with Box Plot Inside')
fig.show()
```

### 3.5 Interactive Heatmap

**Syntax:**
- `px.imshow(data)` - Create interactive heatmap
- Can also use `go.Heatmap()` for more control

```python
# Correlation heatmap using Plotly Express
corr_matrix = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

fig = px.imshow(corr_matrix, 
                text_auto=True,  # Show values
                color_continuous_scale='RdBu_r',
                title='Correlation Heatmap - Iris Dataset')
fig.show()

# Heatmap using Graph Objects (more control)
fig = go.Figure(data=go.Heatmap(
    z=corr_matrix.values,
    x=corr_matrix.columns,
    y=corr_matrix.columns,
    colorscale='Viridis',
    text=corr_matrix.values,
    texttemplate='%{text:.2f}',
    textfont={"size": 12}
))
fig.update_layout(title='Correlation Heatmap (Graph Objects)')
fig.show()
```

### 3.6 3D Plots

Plotly excels at creating interactive 3D visualizations that can be rotated and explored.

**Syntax:**
- `px.scatter_3d(df, x='col1', y='col2', z='col3')` - 3D scatter plot

```python
# 3D Scatter Plot
fig = px.scatter_3d(iris, x='sepal_length', y='sepal_width', z='petal_width',
                    color='species', size='petal_length',
                    title='3D Scatter Plot - Iris Dataset')
fig.show()

# 3D Surface Plot
import numpy as np

# Create data for surface
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale='Viridis')])
fig.update_layout(title='3D Surface Plot',
                  scene=dict(
                      xaxis_title='X',
                      yaxis_title='Y',
                      zaxis_title='Z'
                  ))
fig.show()
```

## 4. Advanced Plotly Features

Plotly offers advanced features for creating complex interactive visualizations.

### 4.1 Subplots in Plotly

```python
# Subplots in Plotly
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Scatter', 'Bar', 'Line', 'Histogram')
)

# Add scatter plot
fig.add_trace(go.Scatter(x=[1, 2, 3], y=[4, 5, 6], mode='markers', name='scatter'),
              row=1, col=1)

# Add bar chart
fig.add_trace(go.Bar(x=['A', 'B', 'C'], y=[10, 15, 13], name='bar'),
              row=1, col=2)

# Add line plot
fig.add_trace(go.Scatter(x=[1, 2, 3], y=[2, 4, 8], mode='lines', name='line'),
              row=2, col=1)

# Add histogram
fig.add_trace(go.Histogram(x=np.random.randn(100), name='histogram'),
              row=2, col=2)

fig.update_layout(height=600, showlegend=True, title_text="Plotly Subplots")
fig.show()
```

### 4.2 Animated Plots

```python
# Animated Plot (Time Series)
df = px.data.gapminder()
fig = px.scatter(df, x="gdpPercap", y="lifeExp", 
                 animation_frame="year", animation_group="country",
                 size="pop", color="continent", hover_name="country",
                 log_x=True, size_max=55, range_x=[100,100000], range_y=[25,90],
                 title="GDP vs Life Expectancy Over Time (Animated)")
fig.show()
```

### 4.3 Custom Layout and Styling

```python
# Custom Layout and Styling
fig = px.scatter(tips_df, x='total_bill', y='tip', color='time')
fig.update_layout(
    title='Custom Styled Plot',
    title_font_size=20,
    title_font_color='darkblue',
    xaxis_title='Total Bill ($)',
    yaxis_title='Tip ($)',
    font=dict(family='Arial', size=12),
    plot_bgcolor='lightgray',
    paper_bgcolor='white'
)
fig.show()
```
